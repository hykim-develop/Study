당신은 편의점 아르바이트생입니다. 손님이 구매한 물품의 총액을 계산하고, 거스름돈을 안내하는 프로그램을 작성하세요.

조건:

물건 가격: 한 개당 1,200원인 삼각김밥을 3개 샀습니다.

지불 금액: 손님이 5,000원 지폐를 냈습니다.

출력 내용: * 총 구매 금액

거스름돈 (지불 금액 - 총 구매 금액)

```python
item_price = 1200
count = 3
pay_money = 5000

#코드 작성
total_price = item_price * count
change = pay_money - total_price

print("총 구매 금액:", total_price, "원")
print("거스름돈:", change, "원")
```

[실습 문제 1] 두 변수의 값 맞바꾸기 (Swap)

파이썬에서 튜플을 활용하면 임시 변수 없이 아주 쉽게 두 변수의 값을 바꿀 수 있습니다.

문제:
변수 a에는 10이, b에는 20이 들어있습니다. 튜플의 원리를 이용하여 두 값을 서로 맞바꾸세요.

```python
a = 10
b = 20

# 코드 작성
# a, b 로 작성시 사실은 (a, b) 인것. 
# (a, b) = (b, a)
a, b = b, a

print("a:", a)
print("b:", b)
```
